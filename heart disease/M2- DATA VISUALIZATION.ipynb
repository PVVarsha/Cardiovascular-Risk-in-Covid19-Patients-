## DATA VISUALIZATION AND DATA ANALYSIS

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('heart_risk.csv')

df.head()
```

```python
df.columns
```

```python
df = df.dropna()
```

```python
df=df.drop_duplicates()
```

```python
df.tail()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
plt.hist(df['Diabetes'],color='purple')

plt.subplot(1,2,2)
plt.hist(df['Heart_Risk'],color='tomato')
```

```python
df.hist(figsize=(15,55),layout=(15,4), color='lightblue')
plt.show()
```

```python
df['Heart_Risk'].hist(figsize=(15,7),color='lightgreen')
```

```python
plt.scatter(df['ECG_Result'],df['Smoking_Status'], color='pink') # scatter, plot, triplot, stackplot
```

```python
sns.boxenplot(df['Smoking_Status'], color='purple') 
# barplot, boxenplot, boxplot, countplot, displot, distplot, ecdfplot, histplot, kdeplot, pointplot, violinplot, stripplot
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
k=['Gender','Smoking_Status','COVID_Severity','ECG_Result','Hypertension','Diabetes','Heart_Risk','Vaccine_Given']

for i in k:
    df[i] = le.fit_transform(df[i]).astype(int)
```

```python
fig,ax=plt.subplots(figsize=(30,20))
sns.heatmap(df.corr(),annot=True,fmt='0.2%',cmap='autumn',ax=ax)
```
